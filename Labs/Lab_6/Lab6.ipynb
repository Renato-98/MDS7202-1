```python
"""
En la siguiente celda se crean los datos ficticios a usar en la sección 1 del lab.
❗No realice cambios a esta celda a excepción de n_samples❗
"""

# Datos a utilizar

# Configuracion
n_samples = 5000 #Este parámetro si lo pueden modificar

def create_data(n_samples):

    # Lunas
    moons = datasets.make_moons(n_samples=n_samples, noise=0.05, random_state=30)
    # Blobs
    blobs = datasets.make_blobs(n_samples=n_samples, random_state=172)
    # Datos desiguales
    transformation = [[0.6, -0.6], [-0.4, 0.8]]
    mutated = (np.dot(blobs[0], transformation), blobs[1])

    # Generamos Dataset
    dataset = {
        'moons':{
            'x': moons[0], 'classes': moons[1], 'n_cluster': 2
        },
        'blobs':{
            'x': blobs[0], 'classes': blobs[1], 'n_cluster': 3
        },
        'mutated':{
            'x': mutated[0], 'classes': mutated[1], 'n_cluster': 3
        }
    }
    return dataset

data_sets = create_data(n_samples)
```

Implementaremos la funcion plot_scatter que aplicara los metodos de clustering requeridos y luego hace los scaterplots de estos.



```python
# Importamos las libreria con los algoritmos de cluster, el silhoutee_score y la libreria time
import time
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score

# Función para aplicar clustering y graficar con Plotly
def plot_scatter(data, n_clusters_dict, title):
    datasets = ['moons', 'blobs', 'mutated']
    algorithms = ['KMeans', 'GMM', 'Ward', 'DBSCAN']

    # Crear subplots 3x4 (3 filas, 4 columnas) con plotly
    fig = make_subplots(rows=3, cols=4, subplot_titles=[f'{alg} - {ds}' for ds in datasets for alg in algorithms])

    for i, dataset_name in enumerate(datasets):
        X = data[dataset_name]['x']
        n_clusters = n_clusters_dict[dataset_name]  # Número de clusters específicos para cada dataset

        # 1. K-Means
        start_time = time.time()
        kmeans = KMeans(n_clusters=n_clusters, random_state=42).fit(X)
        kmeans_labels = kmeans.labels_
        kmeans_silhouette = silhouette_score(X, kmeans_labels)
        kmeans_time = time.time() - start_time

        # 2. GMM
        start_time = time.time()
        gmm = GaussianMixture(n_components=n_clusters, random_state=42).fit(X)
        gmm_labels = gmm.predict(X)
        gmm_silhouette = silhouette_score(X, gmm_labels)
        gmm_time = time.time() - start_time

        # 3. Ward (Agglomerative Clustering)
        start_time = time.time()
        ward = AgglomerativeClustering(n_clusters=n_clusters).fit(X)
        ward_labels = ward.labels_
        ward_silhouette = silhouette_score(X, ward_labels)
        ward_time = time.time() - start_time

        # 4. DBSCAN
        start_time = time.time()
        dbscan = DBSCAN(eps=0.3, min_samples=10).fit(X)
        dbscan_labels = dbscan.labels_
        dbscan_silhouette = silhouette_score(X, dbscan_labels) if len(set(dbscan_labels)) > 1 else -1
        dbscan_time = time.time() - start_time

        # Preparar datos para graficar con plotly
        algorithms_data = [
            (kmeans_labels, kmeans_silhouette, kmeans_time, 'KMeans'),
            (gmm_labels, gmm_silhouette, gmm_time, 'GMM'),
            (ward_labels, ward_silhouette, ward_time, 'Ward'),
            (dbscan_labels, dbscan_silhouette, dbscan_time, 'DBSCAN')
        ]

        for j, (labels, silhouette, exec_time, algo_name) in enumerate(algorithms_data):
            fig.add_trace(go.Scatter(
                x=X[:, 0], y=X[:, 1],
                mode='markers',
                marker=dict(color=labels, colorscale='Viridis', size=5),
                name=f'{algo_name} - {dataset_name}',
                showlegend=False,
                text=[f"Silhouette: {silhouette:.2f}, Time: {exec_time:.2f}s"] * len(X)
            ), row=i+1, col=j+1)

            # Añadir título de cada subplot con tiempo y silhouette
            fig.update_xaxes(title_text=f'{exec_time:.2f} s | Silhouette: {silhouette:.2f}', row=i+1, col=j+1)

    # Ajustes finales
    fig.update_layout(title_text=title, height=900, width=1400, showlegend=False)
    fig.show()

# Parámetros de clustering para cada dataset
n_clusters_dict = {'moons': 2, 'blobs': 3, 'mutated': 2} # Se ve a partir de la imagen original

# Probamos la funcion para el dataset de 1000 samples.
plot_scatter(data_1000, n_clusters_dict, "Comparación de tiempos de ejecución por técnica (Plotly)")
```

*   Las variables `'Age', 'Flight Distance', 'Inflight wifi service', 'Departure/Arrival time convenient', 'Ease of Online booking', 'Gate location', 'Food and drink', 'Online boarding', 'Seat comfort', 'Inflight entertainment', 'On-board service', 'Leg room service', 'Baggage handling','Checkin service', 'Inflight service' y 'Cleanliness'` al tener un rango "pequeño" y no presentar outliers significativos se utilizará **MinMaxScaler**, quedando así en un rango entre 0 y 1.

```python
from sklearn.preprocessing import MinMaxScaler, RobustScaler
from sklearn.compose import ColumnTransformer

# Variables que tienen muchos outliers
variables_con_outliers = ['Departure Delay in Minutes', 'Arrival Delay in Minutes']

# Variables numéricas sin outliers
variables_sin_outliers = df_numericas.columns.difference(variables_con_outliers)

# Crear el transformador combinado usando MinMaxScaler y RobustScaler
scalers = ColumnTransformer(transformers=[
    ('minmax', MinMaxScaler(), variables_sin_outliers),  # MinMaxScaler para las variables sin outliers
    ('robust', RobustScaler(), variables_con_outliers)   # RobustScaler para las variables con outliers
])

# Aplicar los scalers al dataset
df_scaled = scalers.fit_transform(df_numericas)

# Convertir los datos escalados a un DataFrame para mantener las etiquetas originales
df_scaled = pd.DataFrame(df_scaled, columns=variables_sin_outliers.tolist() + variables_con_outliers)

# Ver las primeras filas del dataset escalado
df_scaled.head()
```

```python
# Escriba su código aquí

from sklearn.mixture import GaussianMixture

# Pipeline con GMM
def create_pipeline(n_components):
    pipeline = Pipeline([
        ('gmm', GaussianMixture(n_components=n_components, random_state=42))  # Gaussian Mixture
    ])
    return pipeline
```

```python

# Listas para guardar los resultados
aic_values = []
bic_values = []

# Probar entre 3 y 8 clústers
for n_clusters in range(3, 9):
    # Crear el pipeline con n_clusters
    pipeline = create_pipeline(n_clusters)

    # Ajustar el pipeline a los datos
    pipeline.fit(df_red)

    # Extraer el modelo GMM entrenado
    gmm_model = pipeline.named_steps['gmm']

    # Calcular AIC y BIC
    aic_values.append(gmm_model.aic(df_red))
    bic_values.append(gmm_model.bic(df_red))

# Crear un DataFrame para organizar los resultados
results_df = pd.DataFrame({
    'n_clusters': range(3, 9),
    'AIC': aic_values,
    'BIC': bic_values
})
```

```python
import plotly.express as px

# Graficar AIC y BIC
fig = px.line(results_df, x='n_clusters', y=['AIC', 'BIC'],
              labels={'n_clusters': 'Número de Clústers', 'value': 'Valor'},
              title='Valores de AIC y BIC para diferentes números de clústers')
fig.show()
```

AIC y BIC: Ambos criterios penalizan la complejidad del modelo (más clústers), pero se diferencian en cómo lo hacen. El número óptimo de clústers será aquel que minimice tanto el AIC como el BIC, buscando el punto en el que agregar más clústers no mejora sustancialmente el modelo y sólo aumenta su complejidad.

AIC (Akaike Information Criterion): Mide la calidad del modelo en términos del ajuste a los datos, pero penaliza por el número de parámetros. Un valor más bajo de AIC significa un mejor balance entre ajuste y complejidad del modelo.

BIC (Bayesian Information Criterion): Similar al AIC, pero penaliza más fuertemente el uso de parámetros adicionales, lo que generalmente lleva a modelos más simples.

No hay un valor absoluto de AIC o BIC que sea "bueno" o "malo". Lo importante es la comparación relativa entre los diferentes modelos (con distintos números de clústers). Lo que realmente importa es encontrar el modelo que minimiza los valores de AIC o BIC y
observar si agregar más clústers produce una mejora significativa o si los valores de AIC/BIC se estabilizan o incluso aumentan.

Para este caso probamos podemos ver claramente que para n = 7 clusters obtenemos un aparente minimo local para AIC y BIC, asi que este sera el numero de clusters optimos que elijiremos.
